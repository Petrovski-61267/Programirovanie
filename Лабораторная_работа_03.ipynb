1. **Чем отличается функция с возвратом значения от функций без возврата значения?**
Функция с возвратом значения использует инструкцию `return` вычисляет результат и передает его обратно в ту часть программы, откуда она была вызвана, что позволяет сохранить этот результат в переменную или использовать в дальнейших выражениях. Функция без возврата значения в некоторых языках их называют процедурами выполняет определенный набор действий например, выводит текст на экран, но не возвращает никаких данных в точку вызова в Python по умолчанию такая функция возвращает `None`.
2. **Что такое библиотечная функция?**
Библиотечная функция — это готовая встроенная функция, которая уже написана разработчиками языка Python или создателями сторонних библиотек и входит в стандартный пакет поставки или подключаемые модули. Программист может использовать ее сразу, не заботясь о внутренней реализации например, функции `print`, `len`, `abs` или функции из модуля `random`.


3. **Почему библиотечные функции похожи на "черные ящики"?**
Потому что пользователь программист точно знает, какие входные данные аргументы нужно передать функции и какой результат она вернет, но детали ее внутреннего программного кода скрыты от него. Для работы с функцией не нужно знать, как именно она устроена внутри.


4. **Что делает приведенная ниже инструкция?**
`x = random.randint1, 100`
Эта инструкция генерирует случайное целое число в диапазоне от 1 до 100 включительно с помощью функции `randint` из подключенного модуля `random` и записывает полученное значение в переменную `x`.

`printrandom.randint1, 20`

6. **Для чего используется следующий вызов библиотечной функции?**
`random.seed(20)`
Функция `seed()` задает начальное значение (зерно) для генератора псевдослучайных чисел. Это используется для воспроизводимости результатов: если зафиксировать `seed` определенным числом (например, 20), то при каждом новом запуске программы последовательность «случайных» чисел будет повторяться один в один.


7. **Каким образом функции помогают повторно использовать код в программе?**
Функция позволяет записать блок кода один раз и дать ему имя. После этого программа может обращаться к этому блоку (вызывать его) многократно из разных мест, избегая дублирования кода. Если потребуется изменить логику работы, достаточно внести правки только в тело самой функции, а не во все места ее вызова.


8. **Назовите и опишите две части определения функции.**
Определение функции в Python состоит из:
* **Заголовка (сигнатуры):** начинается с ключевого слова `def`, за которым следует уникальное имя функции, в круглых скобках перечисляются параметры (аргументы), и строка заканчивается двоеточием (`:`).
* **Тела функции:** блок инструкций с отступом, который выполняется каждый раз при вызове функции и описывает ее логику.




9. **Как у функции задать параметры по умолчанию?**
Параметры по умолчанию задаются в заголовке функции путем присваивания значения прямо при объявлении параметра через знак равенства (`=`), например: `def my_func(a, b=10):`. Если при вызове функции аргумент для `b` не передан, функция автоматически использует значение `10`.

10. Что такое **Рекурсия**?
— Это процесс, при котором функция вызывает сама себя (прямо или косвенно).

В программировании рекурсивная функция обычно состоит из двух основных частей:

1. **Базовый случай (условие выхода):** условие, при котором функция перестает вызывать сама себя и возвращает базовый результат. Без него функция будет вызывать себя бесконечно, что приведет к переполнению стека памяти (`RecursionError`).
2. **Рекурсивный шаг:** шаг, на котором функция вызывает себя с новыми аргументами, приближая решение к базовому случаю.

Рекурсия часто используется для задач, которые можно естественным образом разбить на более мелкие задачи того же типа (например, вычисление факториала, обход деревьев или каталогов файлов, алгоритмы вроде быстрой сортировки).
