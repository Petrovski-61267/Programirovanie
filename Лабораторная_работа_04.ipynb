### Ответы на контрольные вопросы (Лабораторная работа №4)

1. Опишите назначение, сходства и различия функций чтения из текстового файла: `read()`, `readline()`, `readlines()`.

* **Назначение и различия:**
* `read()` — считывает весь файл целиком в виде одной большой строки (или до указанного количества символов).


* `readline()` — считывает только одну строку из файла (до символа перевода строки `\n`). При каждом новом вызове считывает следующую строку.


* `readlines()` — считывает весь файл целиком, но разбивает его на список строк, где каждый элемент списка — это отдельная строка файла.




* **Сходства:** Все три метода предназначены для извлечения текстовых данных из открытого файла в память компьютера для их дальнейшей обработки.

2. Как прочитать многострочный текстовый файл не используя явно методы `read()`, `readline()`, `readlines()`?
Многострочный файл можно легко прочитать с помощью прямого обхода самого файлового объекта в цикле `for`. Python автоматически построчно перебирает файл:

file = open('data.txt', 'r')

for line in file:
    print(line)

file.close()


3. Какова задача закрытия файла?
Закрытие файла (метод `close()`) освобождает системные ресурсы и оперативную память, выделенные для работы с этим файлом, а также гарантирует, что все записанные данные корректно сохранятся на жестком диске (буферы сбрасываются). Если файл не закрыть, могут возникнуть утечки памяти или файлы могут заблокироваться для других программ.

4. Что такое позиция считывания файла? Где первоначально находится позиция считывания при открытии файла ввода?

* **Позиция считывания (указатель файла)** — это текущее место (номер байта или символа) в файле, откуда будет производиться следующее чтение или запись.
* При открытии файла для чтения (ввод) позиция считывания всегда автоматически устанавливается в самый **начальный символ** файла (в самую первую позицию, индекс 0).



5. В каком режиме открывается файл, если требуется записать в него данные, но при этом требуется оставить существующее содержимое файла нетронутым? В какую часть файла данные записываются при этом?

* Файл открывается в **режиме дозаписи** (параметр `'a'` — *append*).


* При этом новые данные записываются **в самый конец файла** (существующее до этого содержимое файла остается полностью нетронутым и сдвигается назад).



6. Дайте краткое объяснение, что такое исключение.
Исключение (Exception) — это ошибка, которая возникает во время выполнения программы (в рантайме), когда интерпретатор сталкивается с непредвиденной ситуацией (например, деление на ноль, попытка открыть несуществующий файл или неверный тип данных). Если исключение не обработать, программа аварийно завершит работу.

7. Что происходит, если вызвано исключение, и программа его не обрабатывает инструкцией `try/except`?
Программа аварийно прекращает свое выполнение (падает) и выводит в консоль отчет об ошибке (так называемую трассировку — *Traceback*), с указанием строки кода и типа возникшей проблемы.

8. Какое исключение программа вызывает, когда она пытается открыть несуществующий файл?
Программа вызывает исключение типа **`FileNotFoundError`** (а в более старых версиях Python или при низкоуровневых ошибках ОС — `IOError` / `OSError`).

9. Какое исключение программа вызывает, когда она применяет функцию `float()` для конвертации нечислового строкового значения в число?
Программа вызывает исключение типа **`ValueError`** (ошибка значения).

10. Что происходит, если файл не существует и программа пытается его открыть в режиме дозаписи?
Если файла с таким именем не существует, то при открытии его в режиме дозаписи (`'a'`) Python **автоматически создает новый пустой файл** с указанным именем и уже в него производит запись.
